```python
import time
import pandas as pd

daftar_tahun = range(2019, 2026)
hasil_semua_tahun = []

for tahun in daftar_tahun:

    params = {
        "district_id": "",
        "sub_district_id": "",
        "report_type_id": 431,
        "year": tahun,
        "month": "",
        "date": ""
    }

    try:
        time.sleep(JEDA)

        r = requests.get(
            URL_API,
            params=params,
            headers=HEADERS,
            timeout=TIMEOUT
        )

        if r.status_code == 200:
            hasil = r.json()

            if hasil.get("success"):
                baris = hasil["data"]["original"]["data"]

                print(
                    "Tahun:", tahun,
                    "| Jumlah baris:", len(baris)
                )

                for item in baris:
                    hasil_semua_tahun.append(item)

            else:
                print("Data tidak tersedia:", tahun)

        else:
            print("Gagal:", tahun, r.status_code)

    except Exception as e:
        print("Error tahun", tahun, ":", e)

print("\nTOTAL BARIS:", len(hasil_semua_tahun))
```

```python
import hashlib
import os

def hitung_md5(lokasi_file):
    md5 = hashlib.md5()

    with open(lokasi_file, "rb") as file:
        for bagian in iter(lambda: file.read(4096), b""):
            md5.update(bagian)

    return md5.hexdigest()


# MD5 file Excel asli
md5_excel = hitung_md5(file_provinsi)

# MD5 file CSV hasil pembersihan
md5_csv = hitung_md5(FILE_PROCESSED)

print("=== CHECKSUM MD5 ===")

print("\nFILE ASLI:")
print(file_provinsi)
print("MD5:", md5_excel)

print("\nFILE HASIL PEMBERSIHAN:")
print(FILE_PROCESSED)
print("MD5:", md5_csv)
```

```python

import requests
import pandas as pd
import time

URL_API_BALI = (
    "https://balisatudata.baliprov.go.id/"
    "api/v1/report/subject/data"
)

hasil_tahun = []

for tahun in range(2019, 2026):

    params = {
        "type": "datatables",
        "report_type_id": 1607,
        "year": tahun,
        "month": "",
        "date": ""
    }

    try:
        response = requests.get(
            URL_API_BALI,
            params=params,
            headers=HEADERS,
            timeout=TIMEOUT
        )

        print("TAHUN:", tahun)
        print("STATUS:", response.status_code)

        if response.status_code == 200:
            data = response.json()
            baris = data.get("data", [])

            baris = [
                row for row in baris
                if row[0] != "Total"
            ]

            print("JUMLAH DATA:", len(baris))

            for row in baris:
                hasil_tahun.append({
                    "bulan": row[0],
                    "tahun": row[1],
                    "jumlah_kunjungan": row[2]
                })

        else:
            print("Data tidak berhasil diambil")

    except Exception as e:
        print("ERROR:", e)

    time.sleep(JEDA)

df_wisatawan_bali = pd.DataFrame(hasil_tahun)

print("\nTOTAL SELURUH DATA:", len(df_wisatawan_bali))
display(df_wisatawan_bali.head(15))
```

```python

import os
import hashlib
import pandas as pd
from datetime import date

BASE_DIR = "/content/drive/MyDrive/UTS_DATA_ACQUISITION"

RAW_DIR = os.path.join(BASE_DIR, "data", "raw")
PROCESSED_DIR = os.path.join(BASE_DIR, "data", "processed")
DOCS_DIR = os.path.join(BASE_DIR, "docs")

os.makedirs(DOCS_DIR, exist_ok=True)

file_raw = os.path.join(
    RAW_DIR,
    "kunjungan_wisatawan_bali_api_2021_2024.csv"
)

file_bersih = os.path.join(
    PROCESSED_DIR,
    "kunjungan_wisatawan_bali_api_2021_2024_bersih.csv"
)

file_provenance = os.path.join(
    DOCS_DIR,
    "provenance.csv"
)

def hitung_md5(lokasi):
    with open(lokasi, "rb") as file:
        return hashlib.md5(file.read()).hexdigest()

url_api = (
    "https://balisatudata.baliprov.go.id/"
    "api/v1/report/subject/data"
)

tanggal_akses = date.today().isoformat()

data_baru = [
    {
        "nama_file": os.path.basename(file_raw),
        "sumber": "Satu Data Indonesia Provinsi Bali - API",
        "url": url_api,
        "tanggal_akses": tanggal_akses,
        "lisensi": "Perlu diverifikasi pada portal sumber",
        "jumlah_baris": len(df_wisatawan_bali),
        "jumlah_kolom": len(df_wisatawan_bali.columns),
        "md5": hitung_md5(file_raw),
        "keterangan": (
            "Data mentah API report_type_id 1607, "
            "tahun 2021-2024, HTTP 200"
        )
    },
    {
        "nama_file": os.path.basename(file_bersih),
        "sumber": "Satu Data Indonesia Provinsi Bali - API",
        "url": url_api,
        "tanggal_akses": tanggal_akses,
        "lisensi": "Perlu diverifikasi pada portal sumber",
        "jumlah_baris": len(df_wisatawan_bali_bersih),
        "jumlah_kolom": len(df_wisatawan_bali_bersih.columns),
        "md5": hitung_md5(file_bersih),
        "keterangan": (
            "Data API dibersihkan; jumlah kunjungan "
            "diubah menjadi numerik; tanpa data kosong "
            "dan duplikat"
        )
    }
]

df_baru = pd.DataFrame(data_baru)

if os.path.exists(file_provenance):
    df_lama = pd.read_csv(file_provenance)

    df_lama = df_lama[
        ~df_lama["nama_file"].isin(df_baru["nama_file"])
    ]

    df_provenance = pd.concat(
        [df_lama, df_baru],
        ignore_index=True
    )
else:
    df_provenance = df_baru

df_provenance.to_csv(file_provenance, index=False)

print("PROVENANCE BERHASIL DIPERBARUI")
print("JUMLAH SELURUH CATATAN:", len(df_provenance))
print("LOKASI:", file_provenance)

display(df_provenance.tail(2))
```

```python
# Menggabungkan data Kemenpar dengan referensi BPS

df_gabungan = pd.merge(
    df_provinsi_bersih,
    df_bps_provinsi,
    on="kode_provinsi",
    how="left",
    validate="many_to_one",
    indicator=True
)

print("MERGE BERHASIL")
print("\nJUMLAH BARIS:", len(df_gabungan))
print("JUMLAH KOLOM:", len(df_gabungan.columns))

print("\nSTATUS PENGGABUNGAN:")
print(df_gabungan["_merge"].value_counts())

print("\nDATA HASIL MERGE:")
display(df_gabungan.head(15))

print("\nJUMLAH DATA YANG TIDAK COCOK:")
print((df_gabungan["_merge"] != "both").sum())

print("\nJUMLAH DUPLIKAT SELURUH BARIS:")
print(df_gabungan.duplicated().sum())
```
